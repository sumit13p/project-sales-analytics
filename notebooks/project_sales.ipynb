```python
import pandas as pd

df = pd.read_csv("train.csv")

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%d-%m-%Y"
)

df["Ship Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

print(df.shape)
print(df["Order Date"].dtype)
```

```python
annual_sales = (
    df.groupby(df["Order Date"].dt.year)["Sales"]
    .sum()
)

annual_sales_growth = (
    annual_sales
    .pct_change()
    .mul(100)
    .round(2)
)

print(annual_sales_growth)
```

```python

import pandas as pd

df = pd.read_csv("train.csv")

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%d-%m-%Y"
)

df["Ship Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

print("Dataset shape:", df.shape)
```

```python

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nOrder Date data type:")
print(df["Order Date"].dtype)

print("\nShip Days available:", "Ship Days" in df.columns)
```

```python

# Extract the year from each order date
df["Order Year"] = df["Order Date"].dt.year

# Calculate total Sales for each year
annual_sales = df.groupby("Order Year")["Sales"].sum()

# Calculate year-over-year Sales growth
annual_growth = annual_sales.pct_change() * 100

# Combine the results
annual_sales_summary = pd.DataFrame({
    "Total Sales": annual_sales,
    "YoY Growth %": annual_growth
})

print(annual_sales_summary.round(2))
```

```python

annual_sales_change = annual_sales.diff()

annual_sales_summary["Sales Change"] = annual_sales_change

print(annual_sales_summary.round(2))
```

```python

total_change_from_years = (
    annual_sales.iloc[-1] - annual_sales.iloc[0]
)

sum_of_annual_changes = annual_sales_change.iloc[1:].sum()

print(
    "2015 to 2018 Sales change:",
    round(total_change_from_years, 2)
)

print(
    "Sum of annual Sales changes:",
    round(sum_of_annual_changes, 2)
)

print(
    "Difference:",
    round(total_change_from_years - sum_of_annual_changes, 2)
)
```

```python

starting_sales = annual_sales.iloc[0]
ending_sales = annual_sales.iloc[-1]

overall_growth_pct = (
    (ending_sales - starting_sales)
    / starting_sales
) * 100

print("Starting Sales (2015):", round(starting_sales, 2))
print("Ending Sales (2018):", round(ending_sales, 2))
print("Total Sales increase:", round(ending_sales - starting_sales, 2))
print("Overall Sales growth:", round(overall_growth_pct, 2), "%")
```

```python

number_of_years = annual_sales.index[-1] - annual_sales.index[0]

cagr = (
    (ending_sales / starting_sales) ** (1 / number_of_years) - 1
) * 100

print("Number of years:", number_of_years)
print("Sales CAGR:", round(cagr, 2), "%")
```

```python

annual_sales_summary["CAGR Benchmark %"] = cagr

print(
    annual_sales_summary[
        ["Total Sales", "YoY Growth %", "CAGR Benchmark %"]
    ].round(2)
)
```

```python

annual_sales_summary["Growth Rate Change (pp)"] = (
    annual_sales_summary["YoY Growth %"].diff()
)

print(
    annual_sales_summary[
        ["YoY Growth %", "Growth Rate Change (pp)"]
    ].round(2)
)
```

```python

annual_business_summary = annual_sales_summary[
    [
        "Total Sales",
        "Sales Change",
        "YoY Growth %",
        "Growth Rate Change (pp)"
    ]
].copy()

print(annual_business_summary.round(2))
```

```python

print(
    "Years included:",
    annual_business_summary.index.tolist()
)

print(
    "\nSales totals match:",
    annual_business_summary["Total Sales"].equals(annual_sales)
)

print(
    "Number of annual records:",
    len(annual_business_summary)
)
```
